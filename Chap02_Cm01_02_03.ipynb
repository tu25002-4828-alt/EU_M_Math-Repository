```python
import numpy as np
import math

x = np.random.uniform(0.0, 1.0, 10000)
y = np.random.uniform(0.0, 1.0, 10000)

count = 0

for i in range(10000):

    if math.hypot(x[i], y[i]) <= 1.0:
        count += 1

print(count)

pi = 4 * count / 10000

print(pi)
```
