```python
merge_key = [
    'school','sex','age','address','famsize',
    'Pstatus','Medu','Fedu','Mjob','Fjob',
    'reason','nursery','internet'
]

student_data_merge = pd.merge(
    student_data_math,
    student_data_por,
    on=merge_key,
    how='inner',
    suffixes=('_math', '_por')
)

print(student_data_merge.head())
print(student_data_merge.describe())
```

```python
import matplotlib.pyplot as plt

#数学
plt.hist(student_data_merge['G3_math'])

plt.xlabel('G3_math')
plt.ylabel('count')
plt.grid(True)

plt.show()

#ポルトガル
plt.hist(student_data_merge['G3_por'])

plt.xlabel('G3_por')
plt.ylabel('count')
plt.grid(True)

plt.show()
```

```python
#母親の学歴と数学の成績
plt.scatter(student_data_merge['Medu'],
            student_data_merge['G3_math'])

plt.xlabel('Medu')
plt.ylabel('G3_math')
plt.grid(True)

plt.show()


#父親の学歴と数学の成績
plt.scatter(student_data_merge['Fedu'],
            student_data_merge['G3_math'])

plt.xlabel('Fedu')
plt.ylabel('G3_math')
plt.grid(True)

plt.show()

#数学とポルトガルの成績
plt.scatter(student_data_merge['G3_math'],
            student_data_merge['G3_por'])

plt.xlabel('G3_math')
plt.ylabel('G3_por')
plt.grid(True)

plt.show()
```

考察
G3 のヒストグラム→数学・ポルトガル語ともに中央付近に集中
極端な高得点・低得点は少なく，ポルトガル語の方がやや高得点側に広がっている
Medu / Fedu と成績→学歴が高いほど成績が高い傾向が一部見られるが，強い相関ではなく，ばらつきも大きい。家庭環境だけで成績が決まるとは結論付けられない。
G3_math と G3_por→正の相関が散見される。数学が得意な学生はポルトガル語も比較的高い傾向にあり，「学習習慣や学習意欲など共通要因が考えられる。

数学データ単体との違い

数学データ単体では数学成績しか扱えず，他教科との関連が見えなかった。

一方で，マージ後は同一生徒の複数教科比較が可能かつ「全体的に学力が高い学生か」を見やすくできた。また教科間の相関を分析できた。

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

student_data_por = pd.read_csv('student-por.csv', sep=';')

X = student_data_por[['G1']]
Y = student_data_por['G3']

reg = LinearRegression()

reg.fit(X, Y)

print('回帰係数:', reg.coef_[0])
print('切片:', reg.intercept_)
print('決定係数:', reg.score(X, Y))
```

```python
# 説明変数
X = student_data_por[['G1']]

# 目的変数
Y = student_data_por['G3']

# 回帰モデル
reg = LinearRegression()
reg.fit(X, Y)

Y_pred = reg.predict(X)

# 散布図
plt.scatter(X, Y)

# 回帰直線
plt.plot(X, Y_pred)

plt.xlabel('G1')
plt.ylabel('G3')

plt.grid(True)
plt.show()
```

```python
# 説明変数（欠席数）
X = student_data_por[['absences']]

# 目的変数（最終成績）
Y = student_data_por['G3']

reg = LinearRegression()
reg.fit(X, Y)

# 回帰係数
print('回帰係数:', reg.coef_[0])

# 切片
print('切片:', reg.intercept_)

# 決定係数
print('決定係数:', reg.score(X, Y))

Y_pred = reg.predict(X)

plt.scatter(X, Y)
plt.plot(X, Y_pred)

plt.xlabel('absences')
plt.ylabel('G3')

plt.grid(True)
plt.show()
```

欠席数が増えるほど成績が下がる傾向が見られる場合，回帰係数は負になる。
ただし散布図のばらつきが大きい場合，欠席数だけでは成績を十分説明できない。また，決定係数 $R^2$が小さい場合，欠席数と成績の線形関係は弱いと考えられる
