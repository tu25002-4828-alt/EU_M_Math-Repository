```python
P_D = 0.001

P_pos_given_D = 0.99

P_pos_given_notD = 0.03

P_notD = 1 - P_D

P_D_given_pos =  P_pos_given_D * P_D / (P_pos_given_D * P_D + P_pos_given_notD * P_notD)

print("陽性だった人が病気である確率 =", P_D_given_pos)
print(P_D_given_pos * 100, "%")
```

```python
import numpy as np
import matplotlib.pyplot as plt

ave = 0
disp = 1
N = 100
trial = 10000

sample_means = []

for i in range(trial):
    x = np.random.normal(ave, np.sqrt(disp), N)
    sample_means.append(np.mean(x))


plt.hist(sample_means, bins=30)

plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

ave = 0
disp = 1
N = 100
trial = 10000

sample_means = []

for i in range(trial):
    x = np.random.lognormal(ave, np.sqrt(disp), N)
    sample_means.append(np.mean(x))


plt.hist(sample_means, bins=30)

plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.grid(True)

plt.show()
```

```python
import zipfile
import pandas as pd
from scipy import stats

# CSV読込
student_math = pd.read_csv('student-mat.csv', sep=';')
student_por = pd.read_csv('student-por.csv', sep=';')

# 平均
print('G2平均')
print('Math:', student_math['G2'].mean())
print('Por :', student_por['G2'].mean())

print('\nG3平均')
print('Math:', student_math['G3'].mean())
print('Por :', student_por['G3'].mean())

# Welchのt検定
g2_t, g2_p = stats.ttest_ind(
    student_math['G2'],
    student_por['G2'],
    equal_var=False
)

g3_t, g3_p = stats.ttest_ind(
    student_math['G3'],
    student_por['G3'],
    equal_var=False
)

print('\nG2')
print('t値 =', g2_t)
print('p値 =', g2_p)

print('\nG3')
print('t値 =', g3_t)
print('p値 =', g3_p)

#一般に有意水準5%（0.05）を用いる。
#G2のp値は0.000118であり0.05より十分小さいため、数学とポルトガル語のG2平均には有意な差があるといえる。
#G3のp値も2.22×10⁻⁸であり0.05より十分小さいため、数学とポルトガル語のG3平均にも有意な差があるといえる。
#平均値を見ると、G2・G3ともにポルトガル語の方が数学より高い傾向が見られる。
```

```python
#総合
student_data_math = pd.read_csv('student-mat.csv', sep=';')
student_data_por = pd.read_csv('student-por.csv', sep=';')

student_data_merge = pd.merge(student_data_math
                              ,student_data_por
                              ,on=['school','sex','age','address','famsize','Pstatus','Medu'
                              ,'Fedu','Mjob','Fjob','reason','nursery','internet']
                              ,suffixes=('_math','_por'))

print('G1数学の成績平均：', student_data_merge.G1_math.mean())
print('G1ポルトガル語の成績平均：', student_data_merge.G1_por.mean())

from scipy import stats
t,p = stats.ttest_rel(student_data_merge.G1_math, student_data_merge.G1_por)
print('p値 = ',p)


print('\n --総合問題-- \n')

from scipy import stats

# 欠席数
print('数学の欠席数平均：', student_data_merge.absences_math.mean())
print('ポルトガル語の欠席数平均：', student_data_merge.absences_por.mean())

t, p = stats.ttest_rel(
    student_data_merge.absences_math,
    student_data_merge.absences_por
)

print('欠席数のp値 =', p)

# 勉強時間
print('数学の勉強時間平均：', student_data_merge.studytime_math.mean())
print('ポルトガル語の勉強時間平均：', student_data_merge.studytime_por.mean())

t, p = stats.ttest_rel(
    student_data_merge.studytime_math,
    student_data_merge.studytime_por
)

print('勉強時間のp値 =', p)

#G1の平均点は数学が10.86点，ポルトガル語が12.11点であり，ポルトガル語の方が高かった。
#また，対応のあるt検定の結果、p値は 1.65×10−16となり，有意水準5%を大きく下回った。
#したがって，数学とポルトガル語のG1の平均点には統計的に有意な差があるといえる。
#同じ生徒について比較した場合，ポルトガル語の成績の方が高い傾向あり。

#欠席数について対応のあるt検定を行った結果，p値は0.05未満であった。
#したがって数学とポルトガル語の欠席数には有意な差があるといえる。
```
