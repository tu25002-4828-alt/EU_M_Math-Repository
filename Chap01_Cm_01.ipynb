```python
def prime_number(N):
    for n in range(2, N + 1):
        is_prime = True

        for i in range(2, n):
            if n % i == 0:
                is_prime = False
                break

        if is_prime:
            print(n)

#一般化した素数の出力制限
#prime_number(20)
```
