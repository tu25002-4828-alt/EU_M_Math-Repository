```python
from pandas import Series, DataFrame
import pandas as pd

attri_data1 = {'ID' : ['1','2','3','4','5'],
               'Sex':['F','F','M','M','F'],
               'Money':[1000,2000,500,300,700],
               'Name':['Saito','Horie','Kondo','Kawada','Matsubara']}

attri_data_frame1 = DataFrame(attri_data1)

attri_data2 = {'ID' : ['3','4','7'],
               'Math':[60,30,40],
               'English':[80,20,30]}

attri_data_frame2 = DataFrame(attri_data2)
merge_data = pd.merge(attri_data_frame1,
                      attri_data_frame2,
                      on='ID')

print(merge_data)
print(merge_data[['Money', 'Math', 'English']].mean())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 一様乱数を1000個生成
x = np.random.uniform(0.0, 1.0, 1000)
y = np.random.uniform(0.0, 1.0, 1000)

# グラフ領域を作成
plt.subplot(2, 1, 1)
plt.hist(x)
plt.title("x")

plt.subplot(2, 1, 2)
plt.hist(y)
plt.title("y")

plt.show()

#np.random.uniform(0.0, 1.0, 10)
#np.random.uniform(0.0, 1.0, 100)
#np.random.uniform(0.0, 1.0, 10000)
#個数を増やすと平坦な分布（一様分布）に近づく
```
